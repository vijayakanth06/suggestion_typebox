```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("voice.csv")
print(df.head())
```

```python
df['index'] = np.arange(1, len(df) + 1)
df = df[['index', 'name']]
print(df.head())
```

```python
df['weight'] = df['name'].str.extract(r'-\s*(\d+\s*(GM|ML))', expand=False)[0].fillna('')
df['name'] = df['name'].str.split('-').str[0].str.strip()
print(df.head())
```

```python
df['label'] = df['name'].str.extract(r'\((.*?)\)', expand=False).fillna('')
df['name'] = df['name'].str.replace(r'\s*\(.*?\)', '', regex=True).str.strip()
print(df.head())
```

```python
import re

def extract_weight(text):
    match = re.search(r'(\d+\s*(GM|ML|L|G))', text, re.IGNORECASE)
    return match.group(1) if match else ''

df['weight'] = df.apply(lambda row: extract_weight(row['name']) if extract_weight(row['name']) else row['weight'], axis=1)
df['name'] = df['name'].str.replace(r'\d+\s*(GM|ML|L|G)', '', regex=True).str.strip()
df['weight'] = df.apply(lambda row: extract_weight(row['label']) if extract_weight(row['label']) else row['weight'], axis=1)
df['label'] = df['label'].str.replace(r'\d+\s*(GM|ML|L|G)', '', regex=True).str.strip()

print(df.head())
```

```python
df.to_csv("preprocessed.csv", index=False)
```
